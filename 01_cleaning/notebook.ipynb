## Import library

## Open Dataset

```python
df = pd.read_csv(r'/content/drive/MyDrive/data-analysis-datasets/01_cleaning_datasets/Titanic-Dataset.csv')

display(df.head())
print()
display(df.tail())
```

## Dataset Analysis

```python
print('Titanic Dataset')
print('-' * 50)

print(f"Dataframe shape: {df.shape}")
print('\nDataset info:')
print(df.info())

print(df.describe())
print("\nMissing values:")
print(df.isna().sum())
```

## Filling in the gaps

### Age

```python
df[df['Age'].isna() == True].head()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median()).astype('int64')
```

### Cabin

```python
df[df['Cabin'].isna() == True].head()
```

```python
df['Cabin'] = df['Cabin'].fillna('Unknown')
```

```python
df[df['Cabin'] == 'Unknown'].head()
```

### Embarked

```python
df[df['Embarked'].isna() == True].head()
```

```python
df['Embarked'] = df['Embarked'].fillna('Unknown')
```

## Drop duplicates

```python
display(df[df.duplicated(keep=False) == True].head())  # все вхождения дубликатов
print('\nДубликатов нет')
```

## Change the data type

```python
df['Sex'] = pd.Categorical(df['Sex'])

df['Embarked'] = pd.Categorical(df['Embarked'])
```

```python
df.info()
```

## Save results

```python
df.to_csv(r'/content/drive/MyDrive/data-analysis-datasets/01_cleaning_datasets/cleaned_Titanic_dataset.csv')
```

## Conclusions

 1 Filled in the gaps in the Age column for the median value with the int64 type
