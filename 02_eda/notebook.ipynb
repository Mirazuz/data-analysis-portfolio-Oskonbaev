## Import library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
```

## Dataset country_wise_latest.csv

```python
file_id = '1erCZ_dHFbsO_CyEztXHHMebYTuWeMY11'
url = f'https://drive.google.com/uc?id={file_id}'

df = pd.read_csv(url)
display(df.head())
print()
display(df.tail())
```

## Dataset Analysis

```python
print('COVID-19 Dataset')
print('-' * 50)

print(f"Dataframe shape: {df.shape}")
print('\nDataset info:')
print(df.info())

print(df.describe())
print("\nMissing values:")
print(df.isna().sum())
```

## Drop duplicates

```python
df.drop_duplicates()
```

## Bar chart

```python
CIS_Countries = ['Armenia', 'Azerbaijan', 'Belarus', 'Kazakhstan', 'Kyrgyzstan', 'Moldova', 'Russia', 'Tajikistan', 'Turkmenistan', 'Uzbekistan', 'Ukraine']

CIS_Countries_death = (df
 .query('`Country/Region` in @CIS_Countries')
 ['Deaths / 100 Cases']
 .tolist()
 )

CIS_Countries_dataset = (df
 .query('`Country/Region` in @CIS_Countries')
 ['Country/Region']
 .tolist()
 )

plt.bar(CIS_Countries_dataset, CIS_Countries_death)
plt.title('Процент смертей от числа заболевших в странах СНГ')
plt.ylabel('Процент смертей от числа заболевших')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## Line chart

### Dataset full_grouped.csv

```python
file_id = '1b2tr9P4iSUoBi-RmGBPXxiZGnuFKG4Si'
url = f'https://drive.google.com/uc?id={file_id}'

df_full_grouped = pd.read_csv(url)
display(df_full_grouped.head())
print()
display(df_full_grouped.tail())
```

```python
#Data Analysis

print(f"Dataframe shape: {df_full_grouped.shape}")
print('\nDataset info:')
print(df_full_grouped.info())

print(df_full_grouped.describe())
print("\nMissing values:")
print(df_full_grouped.isna().sum())
```

```python
# drop duplicates
df_full_grouped.drop_duplicates()
```

```python
# Change type from object to datetime
df_full_grouped['Date'] = pd.to_datetime(df_full_grouped['Date'])
```

```python
CIS_Countries = ['Belarus', 'Russia', 'Ukraine']

CIS_Countries_Recovery = (df_full_grouped
                        .query('`Recovered` != 0 and `Country/Region` in @CIS_Countries')
                        )

pivot = CIS_Countries_Recovery.pivot_table(
    index = 'Date',
    columns = 'Country/Region',
    values = 'Recovered',
    aggfunc='max'
)

plt.figure(figsize=(14, 8))
pivot.plot(ax=plt.gca())
plt.title('Динамика выздоровевших в странах СНГ')
plt.ylabel('Число выздоровевших')
plt.xlabel('Дата')
plt.show()
```

## Histogram

```python
plt.hist(df['Deaths / 100 Cases'].dropna(), bins=30, edgecolor='black')
plt.title('Распределение летальности COVID-19 по странам')
plt.xlabel('Deaths / 100 Cases в процентах')
plt.ylabel('Число стран')
plt.tight_layout()
plt.show()
```

## Boxplot

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='WHO Region', y='Deaths / 100 Cases')
plt.xticks(rotation=45)
plt.title('Летальность COVID-19 по регионам WHO')
plt.tight_layout()
plt.show()
```

## Heatmap

```python
plt.figure(figsize=(10, 6))

num_columns = df.select_dtypes('int64').columns

sns.heatmap(
    df[num_columns].corr(),
    annot=True,
    cmap='coolwarm',
    fmt=".2f"
)

plt.title('Тепловая карта корреляции')
plt.show()
```

## Interactive chart

```python
df_full_grouped_Russia = (df_full_grouped
 .query('`Country/Region` == "Russia"')
 )

fig = px.line(df_full_grouped_Russia, x='Date', y='New deaths', template='plotly_dark', title='Ежедневные новые смерти от ковида в России')
fig.update_layout(xaxis_title='Дата', yaxis_title='Новые смерти', hovermode='x unified')
fig.show()
```

## Выводы

Все выводы из графиков основы на датасетах и сильно зависят от полноты тестирования и сбора данных.

Bar char показывает, что Кыргызстан, Молдова и Украина занимают лидирующие места по процентному соотношению смертей от числа заболевших,а самыми низкими показателями обладают Узбекинстан, Казахстан и Беларусь.

Линейный график показывает, что Россия демонстрирует наиболее выраженный рост числа выздоровевших с июня 2020 года. Беларусь и Украина растут медленнее это объясняется меньшим масштабом эпидемии.

По гистограмме видно, что большинство стран имеют летальность от 0 до 3-4 процентов, но также есть страны с летальностью выше 10-15 процентов, это может быть связано с поздним выявлением случаев, слабым уровнем здравоохранения или перегрузкой больниц.

Boxplot показывает большой разброс выбросов в Европе и Африке. Америка имеет самую высокую медиану, что означает, что типичная страна в этом регионе показывает более высокую летальность, чем типичная страна в других регионах.

Тепловая карта подтверждает сильную положительную корреляцию между Confirmed и Deaths, а также между Confirmed и Recovered. Корреляция между Deaths и Recovered слабее, что говорит о том, что эти процессы не всегда идут параллельно. В одних странах умирает больше, в других выздоравливает.

Интерактивный график показывает ежедневные новые смерти в России и к началу июля 2020 года наблюдается снижение этого показателя.
